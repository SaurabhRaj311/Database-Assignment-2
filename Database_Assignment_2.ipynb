#1. What are the key differences between SQL and NoSQL databases?

#**Answer**

Data Structure: SQL databases are relational and store structured data in predefined tables with fixed rows and columns. NoSQL databases are non-relational and store unstructured or semi-structured data using documents, key-value pairs, wide-column stores, or graphs.
- Schema Flexibility: SQL uses a rigid, predefined schema that requires alterations whenever the data model changes. NoSQL features a dynamic schema, allowing fields to be added on the fly without changing existing database structures.
- Scalability: SQL databases scale vertically by adding more computing power (CPU, RAM) to a single server. NoSQL databases scale horizontally by distributing data across multiple commodity servers using sharding.
- Query Language: SQL uses Structured Query Language for defining and manipulating data. NoSQL relies on database-specific object-oriented APIs, document-based queries, or JSON-like syntax.
- Transaction Model: SQL emphasizes ACID properties (Atomicity, Consistency, Isolation, Durability) for reliability. NoSQL typically prioritizes the BASE model (Basically Available, Soft-state, Eventual consistency) to maximize horizontal performance and availability.

#2. What makes MongoDB a good choice for modern applications?

#**Answer**

- Flexible Document Model: Stores data in BSON format (binary JSON), allowing complex hierarchical data models within a single record.
- High Performance: In-memory processing, efficient indexing, and document embedding reduce the need for costly database joins.
- Built-in Scalability: Supports horizontal scaling through native sharding, distributing data across multiple systems easily.
- High Availability: Provides native replication via replica sets, ensuring automatic failover and data redundancy.
- Rich Querying Ability: Supports powerful aggregation frameworks, full-text search, geospatial queries, and indexing capabilities.

#3. Explain the concept of collections in MongoDB.
Definition: A collection is a grouping of related MongoDB documents, directly equivalent to a table in a traditional relational SQL database.
Schema-Less Nature: Unlike SQL tables, collections do not enforce a fixed schema by default. Documents within the same collection can contain different fields or structures.
Storage Organization: Collections reside inside individual databases and store documents in BSON format.

#4. How does MongoDB ensure high availability using replication?
Replica Sets: MongoDB achieves high availability using replica sets, which are groups of nodes consisting of one Primary node and multiple Secondary nodes.
Data Synchronization: The primary node processes all write operations and records them in an operation log (oplog). Secondary nodes continuously copy and apply this oplog to replicate data.
Automatic Failover: If the primary node goes offline, remaining secondary nodes automatically hold an election to promote a new primary node without downtime.

#6. What is the role of indexes in MongoDB, and how do they improve performance?
Role: Indexes create ordered data structures that store a small portion of the collection's dataset in an easily traversable form.
Performance Improvement: Without indexes, MongoDB must perform a collection scan (checking every single document in a collection). Indexes allow MongoDB to instantly locate matching documents, dramatically reducing read query times and disk I/O operations.

#7. Describe the stages of the MongoDB aggregation pipeline.
$match: Filters documents to pass only those that match specified conditions (equivalent to SQL WHERE).
$project: Selects, reshapes, or creates specific fields in the output documents (equivalent to SQL SELECT).
$group: Groups input documents by a specified key and performs aggregations like sum, average, or count (equivalent to SQL GROUP BY).
$sort: Reorders documents based on specified field values (equivalent to SQL ORDER BY).
$limit / $skip: Restricts the number of output documents or skips a specified count for pagination.
$lookup: Performs a left outer join to retrieve related data from another collection.

#8. What is sharding in MongoDB? How does it differ from replication?
Sharding Definition: Sharding is a technique for horizontally partitioning data across multiple independent servers (shards) to distribute heavy storage and read/write workloads.
Difference from Replication: Replication copies identical data across multiple nodes for redundancy and high availability. Sharding splits unique subsets of data across multiple servers to scale capacity and performance.

#9. What is PyMongo, and why is it used?

Purpose: It provides Python developers with simple object-oriented abstractions and methods to create connections, execute queries, run aggregation pipelines, and manipulate MongoDB collections directly within Python scripts.

#10. What are the ACID properties in the context of MongoDB transactions?
Atomicity: Guarantees that all operations within a multi-document transaction complete successfully together, or all changes are completely rolled back if any operation fails.
Consistency: Ensures transactions move the database strictly from one valid state to another, enforcing all schema and domain constraints.
Isolation: Prevents concurrent transactions from interfering with each other; uncommitted changes are invisible to other client operations.
Durability: Ensures that committed transaction modifications are saved permanently in the transaction log (journal) and survive system crashes.

#11. What is the purpose of MongoDB's explain() function?
Purpose: The explain() function returns detailed execution metrics and query planner stats for a specific query.
Key Insights: Helps developers analyze performance bottlenecks by disclosing whether a query utilized an index, total execution time, total documents scanned, and the execution plan selected by MongoDB.

#12. How does MongoDB handle schema validation?
Validation Rules: MongoDB supports JSON Schema validation applied directly at the collection level during collection creation or updates.
Enforcement: Whenever a document is inserted or updated, MongoDB checks the document against defined schema rules (checking required fields, data types, and value ranges).
Action Levels: Rejects non-compliant documents completely if set to error, or permits the operation while logging a compliance flag if set to warn.

#13. What is the difference between a primary and a secondary node in a replica set?
Primary Node: Accepts and processes all write operations from applications, logs them in the oplog, and serves read operations by default. There is only one active primary node per replica set.
Secondary Node: Replicates the primary node's oplog to maintain identical data copies. It cannot accept direct write operations, but can handle read operations if configured and can participate in elections to become the primary node if the current primary fails.

#14. What security mechanisms does MongoDB provide for data protection?
Authentication: Verifies user identities using mechanisms like SCRAM, x.509 certificates, and LDAP integration.
Authorization: Implements Role-Based Access Control (RBAC) to grant granular permissions to specific database resources.
Encryption: Protects data in transit using TLS/SSL and data at rest using storage engine encryption.
Auditing: Enables logging of security events and administrative operations for compliance verification.

#15. Explain the concept of embedded documents and when they should be used.
Concept: Embedded documents are nested JSON/BSON sub-documents placed directly inside a field of a parent document.
When to Use: Use embedded documents for "one-to-one" or bounded "one-to-many" relationships where data is queried together frequently, avoiding joins and ensuring atomic updates.

#16. What is the purpose of MongoDB's $lookup stage in aggregation?
Purpose: Performs a left outer join to combine documents from an unstructured collection with documents from another target collection in the same database.
Mechanism: Matches equality conditions between a field from the input documents and a target field from the joined collection, returning matched items inside a new array field.

#19. How do MongoDB transactions differ from SQL transactions?
Scope: SQL transactions are natively relational across tables. MongoDB historically operated on single-document atomicity and added multi-document ACID transactions later for replica sets and sharded clusters.
Performance Impact: MongoDB multi-document transactions carry higher performance overhead relative to single-document embedded updates; embedding is preferred over multi-document transactions whenever possible.
Concurrency Control: MongoDB uses optimistic concurrency control with snapshot isolation, whereas standard SQL databases often use pessimistic locking mechanisms.

#20. What are the main differences between capped collections and regular collections?
Capped Collections: Fixed-size collections that preserve insertion order. When the pre-allocated memory size limit is reached, older documents are automatically overwritten by newer ones (FIFO behavior). They do not support document deletions.
Regular Collections: Dynamic collections that grow indefinitely based on disk capacity. They support full CRUD operations, individual document deletions, and do not automatically overwrite old records.

#21. What is the purpose of the $match stage in MongoDB's aggregation pipeline?
Purpose: Filters input documents so that only records satisfying specific conditions pass into the subsequent pipeline stages.
Optimization: Placed at the very beginning of an aggregation pipeline to minimize processing volume and leverage database indexes effectively.

#22. How can you secure access to a MongoDB database?
Enable Authentication: Turn on access control so all clients present valid user credentials before connecting.
Role-Based Access Control (RBAC): Restrict user privileges by assigning custom or built-in roles based on least privilege.
Network Restrictions: Configure IP whitelisting, firewalls, and VPC peering to block unauthorized external access.
Enable Encryption: Enforce TLS/SSL for encrypted network communication and turn on storage encryption for data on disk.

#23. What is MongoDB's WiredTiger storage engine, and why is it important?

Importance: Provides document-level concurrency control, checkpointing, fast in-memory caching, and data compression (Snappy/zlib) to optimize storage footprint, memory utilization, and throughput.

```python
!pip install mongomock pandas -q

import mongomock
import pandas as pd
import os

# Updated direct path to your uploaded file in sample_data
file_path = 'sample_data/superstore.csv'

# Fallback check if it got moved to main root directory
if not os.path.exists(file_path):
    file_path = 'superstore.csv'

print(f"Reading file from: {file_path}\n")

# Load CSV file
df = pd.read_csv(file_path, encoding='latin1')

# Connect to mock MongoDB database
client = mongomock.MongoClient()
db = client["SuperstoreDB"]
orders_collection = db["Orders"]

# 1. Insert records into MongoDB
records = df.to_dict(orient="records")
orders_collection.insert_many(records)
print(f"--- Task 1: Loaded {len(records)} records into MongoDB ---")

# 2. Retrieve and print documents (first 3 preview)
print("\n--- Task 2: Retrieve and print documents ---")
for doc in orders_collection.find().limit(3):
    print(doc)

# 3. Count total documents
print("\n--- Task 3: Total number of documents ---")
print(f"Total documents: {orders_collection.count_documents({})}")

# 4. Fetch orders from 'West' region
print("\n--- Task 4: Orders from 'West' region (Count & Sample) ---")
west_count = orders_collection.count_documents({"Region": "West"})
print(f"Total West region orders: {west_count}")
for doc in orders_collection.find({"Region": "West"}).limit(2):
    print(doc)

# 5. Orders where Sales > 500
print("\n--- Task 5: Orders with Sales > 500 (Count & Sample) ---")
high_sales_count = orders_collection.count_documents({"Sales": {"$gt": 500}})
print(f"Orders with Sales > 500: {high_sales_count}")
for doc in orders_collection.find({"Sales": {"$gt": 500}}).limit(2):
    print(doc)

# 6. Top 3 orders with highest Profit
print("\n--- Task 6: Top 3 orders with highest Profit ---")
for doc in orders_collection.find().sort("Profit", -1).limit(3):
    print(doc)

# 7. Update 'First Class' to 'Premium Class'
print("\n--- Task 7: Update Ship Mode ---")
update_res = orders_collection.update_many(
    {"Ship Mode": "First Class"},
    {"$set": {"Ship Mode": "Premium Class"}}
)
print(f"Modified documents count: {update_res.modified_count}")

# 8. Delete orders where Sales < 50
print("\n--- Task 8: Delete orders with Sales < 50 ---")
delete_res = orders_collection.delete_many({"Sales": {"$lt": 50}})
print(f"Deleted documents count: {delete_res.deleted_count}")

# 9. Total sales per region
print("\n--- Task 9: Total Sales per Region ---")
for doc in orders_collection.aggregate([{"$group": {"_id": "$Region", "TotalSales": {"$sum": "$Sales"}}}]):
    print(doc)

# 10. Distinct Ship Mode values
print("\n--- Task 10: Distinct Ship Modes ---")
print(orders_collection.distinct("Ship Mode"))

# 11. Count orders per category
print("\n--- Task 11: Order count per Category ---")
for doc in orders_collection.aggregate([{"$group": {"_id": "$Category", "OrderCount": {"$sum": 1}}}]):
    print(doc)
```
